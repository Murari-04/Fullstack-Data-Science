```python
import os
import warnings
import pandas as pd
warnings.filterwarnings("ignore")
```

```python
from langchain_experimental.agents import create_pandas_dataframe_agent
from langchain_openai import OpenAI
```

```python
os.environ["GEMINI_API_KEY"] = "PASTE API KEY"  
```

```python
df=pd.read_csv(r'C:\Users\Lenovo\Downloads\titanic dataset.csv')
print(df.shape)
df.head()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
llm = ChatGoogleGenerativeAI( model= "gemini-2.5-flash" , temperature=1)
```

```python
agent=create_pandas_dataframe_agent(
    llm, 
    df, 
    verbose=True,
    agent_type = "openai-tools",
    allow_dangerous_code = True
)
```

```python
agent.run("How many rows are there")
```

```python
agent.run("How many people have more than 3 siblings")
```

```python
agent.run( "Which attribute has missing values?" )
```
